```python
import torch

vad = load_silero_vad()
SPEECH_PROB = 0.5

@torch.inference_mode()
def contains_speech(chunk, sr=16000) -> bool:
    """chunk: 512 samples of 16 kHz mono float32 (shape (512,) or (512, 1))"""
    wave = torch.from_numpy(chunk.reshape(-1))
    return vad(wave, sr).item() >= SPEECH_PROB
```
